```python
import tensorflow as tf 
import numpy as np
```

```python
import pandas as pd

data = pd.read_csv("APR_POLY_N20.dat", sep=' ', usecols=['K','F_chi', 'mass', 'lambda'])
print(data)
```

```python
X = data[['K','F_chi', 'mass']]
Y = data[['lambda']]
print(X)
```

```python
print(Y)
```

```python
layers = []
input_layer = tf.keras.layers.Dense(128, input_shape=(3,), activation='relu')
layers.append(input_layer)
for i in range(10):
    layers.append(tf.keras.layers.Dense(64, activation='relu'))
output_layer = tf.keras.layers.Dense(1, activation='sigmoid')
layers.append(output_layer)
model = tf.keras.models.Sequential(layers)

#checkpoint=tf.keras.callbacks.ModelCheckpoint(filepath=save_loc, monitor='val_loss', verbose=1, save_best_only=True,
       # save_weights_only=True, mode='auto', save_freq='epoch', options=None)
lr_adjust=tf.keras.callbacks.ReduceLROnPlateau( monitor="val_loss", factor=0.2, patience=5, verbose=1, mode="auto",
        min_delta=0.001,  cooldown=0,  min_lr=0)

callbacks = [lr_adjust]

model.compile(optimizer=tf.keras.optimizers.Adam(0.01), loss='mse')

```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, Y, test_size=0.33, random_state=42)
```

```python
r = model.fit(X_train, y_train, epochs=10000,callbacks=callbacks, batch_size=100)
```

```python
Yt = model.predict(X_test)
```

```python
print(np.median(Yt - y_test.to_numpy()))
```
